### Worked example: a function that returns a value

This function converts kilometres to miles and returns the result.

```python
def km_to_miles(km):
    return km * 0.621

print(km_to_miles(10))
distance = km_to_miles(42.2)
print(distance)
```

#### Exercise 1: Celsius to Fahrenheit

Write a function `to_fahrenheit(celsius)` that returns `celsius * 9 / 5 + 32`. Print the result for `0`, `25`, and `100`.

```python
def to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32

print(to_fahrenheit(0))
print(to_fahrenheit(25))
print(to_fahrenheit(100))
```

#### Exercise 2: total with tax

Write a function `total_price(price, quantity, tax_rate)` that returns the price times the quantity, plus tax. Test it with `total_price(10, 3, 0.16)`: it should return `34.8`.

```python
def total_price(price, quantity, tax_rate):
    subtotal = price * quantity
    return subtotal + subtotal * tax_rate

print(total_price(10, 3, 0.16))
```

#### Exercise 3: several return values

Write a function `stats(a, b, c)` that returns the total and the average of three numbers. Call it with `4, 8, 9` and print both values.

```python
def stats(a, b, c):
    total = a + b + c
    return total, total / 3

total, average = stats(4, 8, 9)
print(total, average)
```

### Worked example: if / else

This function decides whether a student passed.

```python
def passed(score):
    if score >= 50:
        return "pass"
    else:
        return "fail"

print(passed(73))
print(passed(41))
```

#### Exercise 5: positive, negative, or zero

Write a function `sign(n)` that returns `"positive"`, `"negative"`, or `"zero"`. Test it with `5`, `-2`, and `0`.

```python
def sign(n):
    if n > 0:
        return "positive"
    elif n < 0:
        return "negative"
    else:
        return "zero"

print(sign(5), sign(-2), sign(0))
```

#### Exercise 6: age groups

Write a function `age_group(age)` that returns `"child"` under 13, `"teen"` from 13 to 17, `"adult"` from 18 to 64, and `"senior"` from 65 up. Test it with `8`, `15`, `30`, `70`.

```python
def age_group(age):
    if age < 13:
        return "child"
    elif age < 18:
        return "teen"
    elif age < 65:
        return "adult"
    else:
        return "senior"
```

#### Exercise 7: find the bug

This function should return `"B"` for 80, but it returns `"C"`. Fix it.

```python
def grade(score):
    if score >= 60:
        return "C"
    elif score >= 75:
        return "B"
    elif score >= 90:
        return "A"
    else:
        return "F"

print(grade(80))
```

#### Exercise 8: delivery fee

A shop charges a delivery fee of 3 JD, but delivery is free for orders of 50 JD or more. Write `delivery_fee(order_total)` and print the fee for `20` and `75`.

```python
def delivery_fee(order_total):
    if order_total >= 50:
        return 0
    else:
        return 3

print(delivery_fee(20))
print(delivery_fee(75))
```

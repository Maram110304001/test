### Worked example: whole division and remainder

`//` and `%` together split a quantity into whole units and what is left over. Here: 150 minutes is 2 hours and 30 minutes.

```python
total_minutes = 150
hours = total_minutes // 60
minutes = total_minutes % 60
print(hours, "hours and", minutes, "minutes")
```

#### Exercise 3: splitting into boxes

A shop has 47 items and each box holds 6. Store 47 and 6 in variables, then print how many full boxes you can fill and how many items are left over.

```python
# Write your code here
shop_items = 47
box_size = 6
print("Full boxes:", shop_items // box_size)
print("Left over:", shop_items % box_size)
```

### Question 2

Write a function `is_even(n)` that returns `True` if `n` is even and `False` otherwise. Test it with three numbers.

```python
def is_even(n):
  if n % 2 == 0:
    return True
  else:
    return False

print (is_even(2))
print (is_even(3))
print (is_even(4))
```

### Question 3

Write a function `bmi(weight_kg, height_m)` that returns `weight / height ** 2`. Then write a second function `bmi_category(value)` that returns `"under"` below 18.5, `"normal"` below 25, and `"over"` otherwise. Use both functions on `bmi(70, 1.75)`.

```python
def bmi(weight_kg, height_m):
  return weight_kg / height_m ** 2

def bmi_category(value):
  if value < 18.5:
    return "under"
  elif value < 25:
    return "normal"
  else:
    return "over"

print (bmi_category(bmi(70, 1.75)))
```

### Question 4

Write a function `max_of_three(a, b, c)` that returns the largest of three numbers, using only `if`, `elif`, and `else` (no built-in `max`).

```python
def max_of_three(a, b, c):
  if a > b and a > c:
    return a
  elif b > a and b > c:
    return b
  else:
    return c
print (max_of_three(1, 2, 3))
print (max_of_three(3, 2, 1))
print (max_of_three(1, 3, 2))
```

### Question 6

A parking garage charges 1 JD for the first hour and 0.5 JD for each extra hour, up to a maximum of 5 JD per day. Write `parking_fee(hours)` and print the fee for 1, 4, and 12 hours.

```python
def parking_fee(hours):
  if hours <= 1:
    return 1
  elif hours <= 5:
    return 1 + (hours - 1) * 0.5
  else:
    return 5

print (parking_fee(1))
print (parking_fee(4))
print (parking_fee(12))
```

### Question 3

A rectangle is 12.5 m long and 8 m wide. Store both in variables, then print its area and its perimeter.

```python
length = 12.5
width = 8

area = length * width
perimeter = 2 * (length + width)

print("Area:", int(area), "m²")
print("Perimeter:", int(perimeter), "m")
```

### Question 6

A phone plan costs 15 JD per month. Store the monthly price in a variable and print the cost for one year, then the cost for one year with a 10% discount.

```python
monthly_price = 15

year_cost = monthly_price * 12
discounted_cost = year_cost * 0.9

print("Cost for one year:", year_cost, "JD")
print("Cost for one year with 10% discount:", discounted_cost, "JD")
```
